```python
#a) count similar elements of list and print their index value   
for i in range(20):
    if numbers.count(numbers[i])>1:
        print(numbers[i],"->",end="")
        for j in range(20):
          if numbers[j]==numbers[i]:
            print(j,end="")
        print()
```

```python
#b) count even and odd values of list
even_count=0
odd_count=0

for num in numbers:
    if num%2==0:
        even_count+=1
    else:
        odd_count+=1
print("Even values",even_count)
print("oddd values",odd_count)
```

```python
#c) count positive and negative values of list.
positive_count=0
negative_count=0

for num in numbers:
    if num>0:
        positive_count+=1
    elif num<0:
        negative_count+=1
print("Positive values",positive_count)
print("Negative values",negative_count)
```

```python
#12 An acronym is a word formed by taking the first letters of the words in a phrase and making a word from them. For example, RAM is an acronym for “random access memory.” Write a program that allows the user to type in a phrase and then outputs the acronym for that phrase. Note: the acronym should be all uppercase, even if the words in the phrase are not capitalized.
phrase=input('Enter a phrase')
words=phrase.split()
acronym=""

for letter in words:
    acronym=acronym + letter[0]
    print("Acronym",acronym.upper())
```

```python
#18.Python program to find common elements in two lists.
a=[1,2,2,3,3,4,4,4,5]
b=[2,5,6,2,1,9]
common=[]

for n in a:
    if n in b:
        common.append(n)
print("Common Elements:",common)
```
