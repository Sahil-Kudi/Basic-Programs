```python
n=int(input("Enter a number"))
if n<=1:
    print("The number is not prime")
else:
    for i in range (2,n):
      if n%i==0:
          print("The number is not prime")
          break
    else:
        print("Prime")
```

```python
n=int(input("Enter a number:"))
reverse = 0

while n > 0:
    digit = n%10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reverse =", reverse)
```

```python
year=int(input("Enter a number:"))

if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a leap year")
elif year % 4==0:
    print("Leap Year")
else:
    print("Not a Leap Year")
        
```
