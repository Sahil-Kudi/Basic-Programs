```python
#write a program to check if the element exist in a tuple or not
numbers=(10,20,30,40,50,60,70,80)
element=int(input("Enter a number to search"))
if element in numbers:
    print("The no. exist in the tuple")
else:
    print("The no. doesn't exist in the tuple")
```

```python
#Write a program to convert a list into a tuple
number_list=[1,2,3,4,5,6,7]
number_tuple=tuple(number_list)
print("List:",number_list)
print("Tuple:",number_tuple)
```

```python
#Write a program to find the index of an item in a tuple
numbers=(10,20,30,40,50,60,70,80)
element=int(input("Enter element:"))

if element in numbers:
    print("Index of the number is :",numbers.index(element))
else:
    print("The number is not in  the tuple")
```
